# Read and plot one ATST readout

All conditions in `All_strains`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
import pandas as pd
from ATST import read_atst

study = read_atst("atst_de_melo_2019/atst_de_melo_2019.atst.txt", multi_readouts=True)
readout_id = 'All_strains'
readout = study.readouts[readout_id]
```

```python
readout.plot(group_by=['phage_id', 'pfu_ml'], facet_by=['isolate_id'])
```

Individual source curves are shown; the merged readout repeats control replicate IDs within Lfar01 conditions.


```python
with pd.option_context("display.max_rows", None, "display.max_columns", None, "display.max_colwidth", None):
    for name, block in [("FILE_INFO", study.file_info), ("STUDY", study.study),
                        ("METADATA", readout.metadata), ("ASSAY", readout.assay)]:
        print(name)
        display(block.data)
    print("LAYOUT")
    display(readout.layout.data)
    for name, table in study.entities.tables.items():
        print(name)
        display(table.data)
```
