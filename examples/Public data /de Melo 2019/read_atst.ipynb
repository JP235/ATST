# Read and plot one ATST readout

All conditions in `All_strains`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
from ATST import read_atst

study = read_atst("atst_de_melo_2019/atst_de_melo_2019.atst.txt", )
```

```python
study.plot(
    group_by=["phage_id", "pfu_ml"],
    facet_by=["isolate_id"],
    summary="median_minmax",
    points_only=False,
)
```

Individual source curves are shown; the merged readout repeats control replicate IDs within Lfar01 conditions.


```python
for name, block in [
    ("FILE_INFO", study.file_info),
    ("STUDY", study.study),
    ("METADATA", study.metadata),
    ("ASSAY", study.assay),
]:
    print(name)
    display(block.data)
print("LAYOUT")
display(study.layout.data)
for name, table in study.entities.tables.items():
    print(name)
    display(table.data)
```
