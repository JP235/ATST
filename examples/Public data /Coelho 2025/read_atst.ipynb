# Read and plot one ATST readout

All conditions in `Fig_1A`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
import pandas as pd
from ATST import read_atst

study = read_atst("atst_coelho_2025/atst_coelho_2025.atst.txt")
```

```python
study.plot(
    group_by=["phage_id"],
    summary="median_minmax",
    labels={"": "PAO1 Control", "VAC1": "PAO1 + VAC1", "VAC3": "PAO1 + VAC3"},
)
```

```python
with pd.option_context(
    "display.max_rows", None, "display.max_columns", None, "display.max_colwidth", None
):
    for name, block in [
        ("FILE_INFO", study.file_info),
        ("STUDY", study.study),
        ("METADATA", study.metadata),
        ("ASSAY", study.assay),
    ]:
        print(name)
        display(block.data)
    print("LAYOUT")
    display(study.layout.data)
    for name, table in study.entities.tables.items():
        print(name)
        display(table.data)
```

```python
display(study.summarize(group_by=["phage_id"]))
```
