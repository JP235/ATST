# Read and plot one ATST readout

All conditions in `MOI_controlled`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
from ATST import read_atst

study = read_atst("atst_oka_2026/atst_oka_2026.atst.txt", multi_readouts=True)
readout_id = "MOI_controlled"
readout = study.readouts[readout_id]
```

```python
readout.plot(group_by=["moi"], filters={"isolate_id":'E_coli_NBRC13168'}, facet_by='phage_id')
```

```python
for name, block in [
    ("FILE_INFO", study.file_info),
    ("STUDY", study.study),
    ("METADATA", readout.metadata),
    ("ASSAY", readout.assay),
]:
    print(name)
    display(block.data)
print("LAYOUT")
display(readout.layout.data)
for name, table in study.entities.tables.items():
    print(name)
    display(table.data)
```
