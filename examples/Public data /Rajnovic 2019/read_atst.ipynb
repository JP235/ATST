# Read and plot one ATST readout

All conditions in `CFU_10e5`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
import pandas as pd
from ATST import read_atst

study = read_atst("atst_rajnovic_2019/atst_rajnovic_2019.atst.txt", multi_readouts=True)
readout_id = "CFU_10e5"
readout = study.readouts[readout_id]
```

```python
fig = readout.plot(group_by=["pfu_ml"])
```

```python
readout.plot(group_by=["pfu_ml"], summary="median_minmax")
```

```python
for name, block in [
    ("FILE_INFO", study.file_info),
    ("STUDY", study.study),
    ("METADATA", readout.metadata),
    ("ASSAY", readout.assay),
]:
    print(name)
    display(block.data)
print("LAYOUT")
display(readout.layout.data)
for name, table in study.entities.tables.items():
    print(name)
    display(table.data)
```

```python
display(readout.summarize(group_by=["pfu_ml"]))
```
