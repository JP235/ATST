# Read and plot one ATST readout

All conditions in `Fig_S6B`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
import pandas as pd
from ATST import read_atst

study = read_atst("atst_burkle_2025/atst_burkle_2025.atst.txt", multi_readouts=True)
readout_id = "Fig_S6B"
readout = study.readouts[readout_id]
```

```python
readout.plot(group_by=["treatment"])
```

```python
readout.plot(group_by=["treatment"], summary="median_minmax")
```

```python
for name, block in [
    ("FILE_INFO", study.file_info),
    ("STUDY", study.study),
    ("METADATA", readout.metadata),
    ("ASSAY", readout.assay),
]:
    print(name)
    display(block.data)
print("LAYOUT")
display(readout.layout.data)
for name, table in study.entities.tables.items():
    print(name)
    display(table.data)
```

```python
display(readout.summarize(group_by=["treatment"]))
```
