# Read and plot one ATST readout

All conditions in `Fig_2B`; labels, colours and units follow the ATST data.


```python
from IPython.display import display
import pandas as pd
from ATST import read_atst

study = read_atst("atst_li_2025/atst_li_2025.atst.txt", multi_readouts=True)
readout_id = 'Fig_2B'
readout = study.readouts[readout_id]
```

```python
readout.plot(group_by=['phage_id', 'moi'], facet_by=['signaling_molecules'])
```

```python
readout.plot(group_by=['phage_id', 'moi'], facet_by=['signaling_molecules'], summary="median_minmax")
```

```python
with pd.option_context("display.max_rows", None, "display.max_columns", None, "display.max_colwidth", None):
    for name, block in [("FILE_INFO", study.file_info), ("STUDY", study.study),
                        ("METADATA", readout.metadata), ("ASSAY", readout.assay)]:
        print(name)
        display(block.data)
    print("LAYOUT")
    display(readout.layout.data)
    for name, table in study.entities.tables.items():
        print(name)
        display(table.data)
```

```python
display(readout.summarize(group_by=['signaling_molecules', 'phage_id', 'moi']))
```
